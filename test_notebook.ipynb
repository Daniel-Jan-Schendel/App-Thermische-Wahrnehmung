# EU AI Act verstehen

## 🎯 **Was ist der EU AI Act?**

Der **EU AI Act** ist das weltweit erste umfassende Gesetz zur Regulierung Künstlicher Intelligenz. Er trat 2024 in Kraft und definiert klare Regeln für KI-Systeme in der Europäischen Union.

### 🚦 **Das Ampel-System: 4 Risikokategorien**

Der EU AI Act funktioniert wie eine **Ampel** - je höher das Risiko, desto strenger die Regeln:

#### 🔴 **Unannehmbares Risiko - VERBOTEN**
- 🚫 **Soziale Bewertungssysteme** (Social Scoring wie in China)
- 🚫 **Manipulative KI-Systeme** (die Menschen unbewusst beeinflussen)
- 🚫 **Biometrische Überwachung** in Echtzeit in öffentlichen Räumen
- 🚫 **Ausnutzung von schutzbedürftige Menschen** (Kinder, Behinderte)

#### 🟠 **Hohes Risiko - STRENGE REGELN**
- 🏥 **Medizinische Geräte** (KI-gestützte Diagnosen)
- 🚗 **Kritische Infrastruktur** (Verkehr, Energie, Wasser)
- 🎓 **Bildungswesen** (Prüfungsbewertung, Zulassungen)
- 👔 **Personalwesen** (Bewerbungsauswahl, Leistungsbewertung)
- 🏛️ **Justizsystem** (Risikoeinschätzung bei Straftätern)

#### 🟡 **Begrenztes Risiko - TRANSPARENZPFLICHT**
- 🤖 **Chatbots** (müssen sich als KI zu erkennen geben)
- 🎭 **Deepfakes** (müssen als künstlich generiert gekennzeichnet werden)
- 👁️ **Emotionserkennung** (Nutzer müssen informiert werden)

#### 🟢 **Minimales Risiko - FREIWILLIG**
- 🎮 **Spiele-KI**
- 📧 **Spam-Filter**
- 🛒 **Produkt-Empfehlungen**

---

## 🎯 **Warum ist das wichtig für Entwickler?**

### 💼 **Praktische Auswirkungen:**
- **📋 Dokumentationspflicht**: Alle KI-Entscheidungen müssen nachvollziehbar sein
- **🔍 Transparenz**: Nutzer müssen verstehen, wie KI-Systeme funktionieren
- **⚖️ Rechtliche Haftung**: Bei Verstößen drohen hohe Geldstrafen (bis 7% des Jahresumsatzes!)
- **🔄 Kontinuierliche Überwachung**: KI-Systeme müssen permanent überprüft werden

# Warum Explainable AI?

## 🤔 **Das Problem: Black Box AI**

Stell dir vor, du bekommst einen Kredit abgelehnt und fragst nach dem Grund:

**Traditionelle Antwort**: *"Der Computer hat 'Nein' gesagt."* 

**Mit Explainable AI**: *"Aufgrund Ihres Einkommens (30%), Ihrer Kredithistorie (45%) und der Antragshöhe (25%) wurde der Antrag abgelehnt."* ✅

### 🧠 **Das Black Box Problem**

Moderne KI-Systeme sind wie **magische Boxen**:
- 📥 **Input**: Daten gehen rein
- ⚫ **Black Box**: Komplexe Berechnungen
- 📤 **Output**: Entscheidung kommt raus

**Problem**: Wir wissen nicht, **WARUM** die KI eine bestimmte Entscheidung getroffen hat!

---

## ⚖️ **EU AI Act: Transparenz ist Pflicht**

#### 📖 **Artikel 13: Transparenz und Nutzungsinformationen**

**Was verlangt das Gesetz?**
Hochrisiko-KI-Systeme müssen **so transparent** gestaltet werden, dass Nutzer die Ausgaben **interpretieren** und **angemessen verwenden** können.

**Konkrete Anforderungen:**
- ✅ **Klare Dokumentation**: Wie funktioniert das System?
- ✅ **Leistungsinformationen**: Genauigkeit, Fehlerrate, Grenzen
- ✅ **Interpretationshilfen**: Wie soll man die Ausgaben verstehen?
- ✅ **Risikoinformationen**: Wo kann das System versagen?
- ✅ **Wartungshinweise**: Wie bleibt das System korrekt?

#### 📖 **Artikel 50: Informationspflichten für alle KI-Systeme**

**Kernprinzip**: Menschen haben das **Recht zu wissen**, wann sie mit KI interagieren.

**Praktische Umsetzung:**
- 🤖 **Chatbots**: "Sie sprechen mit einem KI-Assistenten."
- 🎭 **Deepfakes**: "Dieses Video wurde von KI generiert."
- 😊 **Emotionserkennung**: "Diese App analysiert Ihre Emotionen."
- 👁️ **Biometrische Systeme**: "Ihr Gesicht wird zur Identifikation verwendet."

---

## 🛠️ **Explainable AI (XAI) - Die Lösung**

**Explainable AI** macht KI-Entscheidungen **transparent** und **nachvollziehbar**.

### 🎯 **Zwei Hauptansätze:**

#### 🌍 **1. SHAP (SHapley Additive exPlanations)**
- **Global**: Erklärt das **gesamte** Modell
- **Mathematisch fundiert**: Basiert auf Spieltheorie
- **Umfassend**: Zeigt sowohl positive als auch negative Einflüsse

#### 🔍 **2. LIME (Local Interpretable Model-agnostic Explanations)**
- **Lokal**: Erklärt **einzelne** Vorhersagen
- **Einfach**: Nutzt einfache, verständliche Modelle
- **Praktisch**: Perfekt, um einzelne Entscheidungen zu erklären

---

## 💡 **Warum ist das geschäftskritisch?**

### 🏢 **Business-Vorteile von XAI:**
- **💰 Risikominimierung**: Rechtliche Compliance, weniger Haftungsrisiken
- **🤝 Vertrauen schaffen**: Kunden verstehen und akzeptieren KI-Entscheidungen
- **🔧 Modell-Verbesserung**: Schwächen erkennen und beheben
- **📈 Bessere Performance**: Bias erkennen und Fairness sicherstellen
- **🚀 Marktvorteil**: Frühe Compliance als Wettbewerbsvorteil

### ⚠️ **Risiken ohne XAI:**
- **💸 Geldstrafen**: Bis zu 7% des Jahresumsatzes (für große Unternehmen Millionen!)
- **😡 Kundenverlust**: Intransparente Systeme führen zu Misstrauen
- **📉 Diskriminierung**: Unbewusste Voreingenommenheit (Bias) bleibt unentdeckt
- **🚫 Marktausschluss**: Ohne Compliance kein Marktzugang in der EU

**Fazit**: Explainable AI ist nicht optional - es ist **überlebenswichtig** für moderne (KI-)Unternehmen!

````python
#  Explorative Datenanalyse - Schritt für Schritt

# Bibliotheken importieren mit Erklärungen
print("Importiere notwendige Bibliotheken...")
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
import warnings
warnings.filterwarnings("ignore")

# Einheitliches Design für professionelle Plots
plt.style.use('default')
plt.rcParams['figure.facecolor'] = 'white'
plt.rcParams['axes.facecolor'] = 'white'
plt.rcParams['font.size'] = 10
plt.rcParams['axes.titlesize'] = 14
plt.rcParams['axes.labelsize'] = 11

# Konsistente Farben
farben = {
    'normal': '#2E86AB',      # Professionelles Blau
    'betrug': '#F24236',      # Warnendes Rot
    'akzent': '#F6AE2D'       # Aufmerksamkeits-Gelb
}

print("Bibliotheken erfolgreich geladen!")

# Dataset laden mit Fehlerbehandlung
print("Lade Kreditkarten-Transaktions-Dataset...")
try:
    df_betrug = pd.read_csv('creditcard.csv')
    print("Dataset erfolgreich geladen!")
except FileNotFoundError:
    print("Fehler: creditcard.csv nicht gefunden!")
    print("Stelle sicher, dass die Datei im gleichen Ordner liegt")
    exit()

# Grundlegende Dataset-Informationen
print("\n" + "="*50)
print("DATASET ÜBERSICHT")
print("="*50)
print(f"Anzahl Zeilen: {df_betrug.shape[0]:,}")
print(f"Anzahl Spalten: {df_betrug.shape[1]:,}")
print(f"Speicherbedarf: {df_betrug.memory_usage(deep=True).sum() / 1024**2:.1f} MB")
print(f"Fehlende Werte: {df_betrug.isnull().sum().sum():,}")

# Zielvariable umbenennen
target_col = 'Betrug' if 'Betrug' in df_betrug.columns else 'Class'
if target_col == 'Class':
    df_betrug = df_betrug.rename(columns={'Class': 'Betrug'})
    print("Zielvariable von 'Class' zu 'Betrug' umbenannt")

# Klassenverteilung analysieren (kritisch für EU AI Act!)
betrug_stats = df_betrug['Betrug'].value_counts()
normal_count = betrug_stats[0]
betrug_count = betrug_stats[1]
betrug_rate = betrug_count / len(df_betrug) * 100

print(f"\nKLASSENVERTEILUNG (kritisch für Bias-Bewertung)")
print("="*50)
print(f"Normale Transaktionen: {normal_count:,} ({100-betrug_rate:.2f}%)")
print(f"Betrügerische Transaktionen: {betrug_count:,} ({betrug_rate:.2f}%)")
print(f"Unbalance-Ratio: {normal_count/betrug_count:.1f}:1")

# EU AI Act Compliance Check
if betrug_rate < 1.0:
    print(f"WARNUNG: Stark unbalancierte Daten!")
    print(f"EU AI Act Compliance: Bias-Risiko dokumentieren")
    print(f"Empfehlung: Spezielle Sampling-Techniken verwenden")

print("\nErstelle umfassende Datenvisualisierung...")

# 4-Panel Visualisierung
fig, axes = plt.subplots(2, 2, figsize=(18, 14))
fig.suptitle('Explorative Datenanalyse', 
             fontsize=18, fontweight='bold', y=0.96)

# 1. Klassenverteilung - für EU AI Act Bias-Assessment
print("Panel 1: Klassenverteilung...")
bars = axes[0,0].bar(['Normal', 'Betrug'], betrug_stats.values, 
                     color=[farben['normal'], farben['betrug']], 
                     alpha=0.8, edgecolor='black', linewidth=1)
axes[0,0].set_title('Klassenverteilung', fontweight='bold', pad=15)
axes[0,0].set_ylabel('Anzahl Transaktionen', fontweight='bold')

# Werte auf Balken anzeigen - mit verbesserter Positionierung
for i, (bar, value) in enumerate(zip(bars, betrug_stats.values)):
    height = bar.get_height()
    axes[0,0].text(bar.get_x() + bar.get_width()/2., height + height*0.02,
                   f'{value:,}\n({value/len(df_betrug)*100:.1f}%)',
                   ha='center', va='bottom', fontweight='bold', fontsize=10)
    
axes[0,0].grid(axis='y', alpha=0.5)
axes[0,0].set_ylim(0, max(betrug_stats.values) * 1.15)

# 2. Betragverteilung
print("Panel 2: Betragverteilung...")
df_normal = df_betrug[df_betrug['Betrug'] == 0]['Amount']
df_betrug_data = df_betrug[df_betrug['Betrug'] == 1]['Amount']

bp = axes[0,1].boxplot([df_normal, df_betrug_data], 
                       labels=['Normal', 'Betrug'],
                       patch_artist=True, widths=0.6)
bp['boxes'][0].set_facecolor(farben['normal'])
bp['boxes'][1].set_facecolor(farben['betrug'])
bp['boxes'][0].set_alpha(0.8)
bp['boxes'][1].set_alpha(0.8)

axes[0,1].set_title('Transaktionsbeträge', fontweight='bold', pad=15)
axes[0,1].set_ylabel('Betrag (€)', fontweight='bold')
axes[0,1].set_yscale('log')  # Log-Skala für bessere Sichtbarkeit
axes[0,1].grid(alpha=0.5)
axes[0,1].tick_params(axis='x', labelsize=10)

# 3. Top Features - Feature Importance
print("Panel 3: Feature-Korrelationen...")
correlations = df_betrug.corr()['Betrug'].abs().sort_values(ascending=False)
top_features = correlations.head(8)[1:]  # Top 7 Features (ohne Betrug selbst)

bars = axes[1,0].barh(range(len(top_features)), top_features.values, 
                      color=farben['akzent'], alpha=0.8, edgecolor='black', linewidth=1)
axes[1,0].set_yticks(range(len(top_features)))
axes[1,0].set_yticklabels(top_features.index, fontweight='bold', fontsize=9)
axes[1,0].set_xlabel('Absolute Korrelation mit Betrug', fontweight='bold')
axes[1,0].set_title('Stark korrelierte Prädiktoren', fontweight='bold', pad=15)
axes[1,0].grid(axis='x', alpha=0.5)

# Korrelationswerte anzeigen - verbesserte Positionierung
for i, (bar, value) in enumerate(zip(bars, top_features.values)):
    width = bar.get_width()
    axes[1,0].text(width + 0.005, bar.get_y() + bar.get_height()/2.,
                   f'{value:.3f}', ha='left', va='center', fontweight='bold', fontsize=8)

# 4. Zeitverteilung - Temporal Bias Check
print("Panel 4: Zeitliche Verteilung...")
time_hours = df_betrug['Time'] / 3600  # Sekunden zu Stunden
normal_time = time_hours[df_betrug['Betrug'] == 0]
betrug_time = time_hours[df_betrug['Betrug'] == 1]

axes[1,1].hist(normal_time, bins=48, alpha=0.7, 
               color=farben['normal'], label='Normal', density=True, edgecolor='black', linewidth=0.5)
axes[1,1].hist(betrug_time, bins=48, alpha=0.7, 
               color=farben['betrug'], label='Betrug', density=True, edgecolor='black', linewidth=0.5)
axes[1,1].set_xlabel('Zeit (Stunden seit Start)', fontweight='bold')
axes[1,1].set_ylabel('Relative Häufigkeit', fontweight='bold')
axes[1,1].set_title('Zeitliche Verteilung', fontweight='bold', pad=15)
axes[1,1].legend(loc='upper right', fontsize=10)
axes[1,1].grid(True, alpha=0.3)

# Verbesserte Subplot-Abstände
plt.subplots_adjust(left=0.08, bottom=0.08, right=0.95, top=0.92, wspace=0.25, hspace=0.35)
plt.show()

print("\n" + "="*50)
print("STATISTISCHE ANALYSE")
print("="*50)

# Detaillierte Statistiken für EU AI Act Dokumentation
print(f"BETRAG-STATISTIKEN:")
print(f"Normal  - Mittelwert: {df_normal.mean():.2f} €, Median: {df_normal.median():.2f} €")
print(f"Betrug  - Mittelwert: {df_betrug_data.mean():.2f} €, Median: {df_betrug_data.median():.2f} €")
print(f"Verhältnis: {df_normal.mean()/df_betrug_data.mean():.1f}x höhere Normalbeträge")

print(f"\nTOP BETRUGS-INDIKATOREN:")
for i, (feature, corr) in enumerate(top_features.items(), 1):
    print(f"   {i}. {feature}: {corr:.3f} Korrelation")

print(f"\nZEITLICHE MUSTER:")
normal_peak_hour = (normal_time.value_counts().index[0])
betrug_peak_hour = (betrug_time.value_counts().index[0]) 
print(f"Normal-Peak: Stunde {normal_peak_hour:.0f}")
print(f"Betrug-Peak: Stunde {betrug_peak_hour:.0f}")

print(f"\nEU AI ACT COMPLIANCE STATUS:")
print("Datenqualität: Vollständig, keine fehlenden Werte")
print("Bias-Assessment: Unbalance erkannt und dokumentiert")
print("Feature-Analyse: Mögliche relevante Prädiktoren identifiziert")
print("Zeitlicher Bias: Keine extreme zeitliche Verzerrung")
print("Action Items: Sampling-Strategie für Unbalance entwickeln")

print("\nBereit für Modell-Entwicklung!")
print("="*70)
````

**Besonderheit**: Wir verwenden `class_weight='balanced'` um das **Unbalance-Problem** zu adressieren.

````python
import numpy as np
import pandas as pd
import seaborn as sns
import time
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.ensemble import RandomForestClassifier
import matplotlib.pyplot as plt
from sklearn.metrics import (accuracy_score, confusion_matrix, classification_report, 
                           precision_recall_curve, roc_curve, auc)

# Konsistente Farben für professionelle Visualisierungen
farben = {
    'normal': '#2E86AB',      # Professionelles Blau  
    'betrug': '#F24236',      # Warnendes Rot
    'akzent': '#F6AE2D',      # Aufmerksamkeits-Gelb
    'erfolg': '#A7C957'       # Erfolgs-Grün
}

print("=== RANDOM FOREST KLASSIFIKATION FÜR BETRUGS-ERKENNUNG ===")
print("Entwicklung eines EU AI Act-konformen KI-Systems mit umfassender Erklärbarkeit")
print()

# Datenaufbereitung für Random Forest
print("SCHRITT 1: DATENAUFBEREITUNG")
print("-" * 40)

target_col = 'Betrug'  # Korrekter Spaltenname
X = df_betrug.drop([target_col], axis=1)
y = df_betrug[target_col]
feature_names = X.columns.tolist()

# Train-Test Split mit stratifizierter Aufteilung, gleiche Klassenverhältnisse in Train und Test
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42, stratify=y
)

print(f"Datensatz aufgeteilt:")
print(f"Training: {X_train.shape[0]:,} Transaktionen")
print(f"Test: {X_test.shape[0]:,} Transaktionen")
print(f"Features: {X_train.shape[1]} Merkmale")

# Klassenverteilung analysieren (kritisch für EU AI Act!)
train_class_dist = dict(zip(*np.unique(y_train, return_counts=True)))
test_class_dist = dict(zip(*np.unique(y_test, return_counts=True)))

print(f"\nSCHRITT 2: KLASSENVERTEILUNG (EU AI Act Bias-Check)")
print("-" * 40)
print(f"Training: Legitim = {train_class_dist[0]:,} ({train_class_dist[0]/len(y_train)*100:.2f}%), " + 
      f"Betrug = {train_class_dist[1]:,} ({train_class_dist[1]/len(y_train)*100:.2f}%)")
print(f"Test: Legitim = {test_class_dist[0]:,} ({test_class_dist[0]/len(y_test)*100:.2f}%), " + 
      f"Betrug = {test_class_dist[1]:,} ({test_class_dist[1]/len(y_test)*100:.2f}%)")
print(f"Imbalance-Ratio: 1:{train_class_dist[0]/train_class_dist[1]:.0f}")
print("EU AI Act: Extreme Unbalance erkannt → class_weight='balanced' erforderlich")

# Feature-Normalisierung (wichtig für Random Forest Performance)
print(f"\nSCHRITT 3: FEATURE-NORMALISIERUNG")
print("-" * 40)
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
print("StandardScaler angewendet (verhindert Data Leakage)")
print("EU AI Act: Korrekte Trennung von Training/Test-Preprocessing")

# Random Forest Training mit EU AI Act Compliance
print(f"\nSCHRITT 4: RANDOM FOREST TRAINING")
print("-" * 40)
print("Modell-Konfiguration für EU AI Act Compliance:")
print("Interpretierbarkeit: Random Forest (natürlich erklärbar)")
print("Fairness: class_weight='balanced' für Unbalance-Korrektur")
print("Robustheit: Begrenzte Tiefe gegen Overfitting")
print("Reproduzierbarkeit: Fixed random_state")

start_time = time.time()
rf_model = RandomForestClassifier(
    n_estimators=100,        # Mehr Bäume für stabilere Vorhersagen
    max_depth=10,            # Begrenzte Tiefe gegen Overfitting
    min_samples_split=5,     # Mindest-Samples für Split
    min_samples_leaf=2,      # Mindest-Samples pro Blatt
    bootstrap=True,          # Bootstrap für Robustheit
    random_state=42,         # Reproduzierbarkeit
    class_weight='balanced', # EU AI Act: Fairness durch Gewichtung / alternativ: Oversampling
    n_jobs=-1                # Parallelisierung für Performance
)

rf_model.fit(X_train_scaled, y_train)
training_time = time.time() - start_time
print(f"Training abgeschlossen in {training_time:.2f} Sekunden")

# Modell-Evaluation mit umfassenden Metriken
print(f"\nSCHRITT 5: MODELL-EVALUATION")
print("-" * 40)

# Vorhersagen
y_pred = rf_model.predict(X_test_scaled)
y_pred_proba = rf_model.predict_proba(X_test_scaled)[:, 1]

# Basis-Metriken
accuracy = accuracy_score(y_test, y_pred)
print(f"Accuracy: {accuracy:.4f} ({accuracy*100:.2f}%)")

# EU AI Act relevante Metriken
cm = confusion_matrix(y_test, y_pred)
tn, fp, fn, tp = cm.ravel()
print(f"\nEU AI ACT RELEVANTE METRIKEN:")
print(f"True Positives (Betrug erkannt): {tp}")
print(f"False Positives (fälschlicherweise Betrug): {fp}")
print(f"False Negatives (Betrug übersehen): {fn}")
print(f"True Negatives (korrekt als legitim): {tn}")
print(f"Precision (Betrug): {tp/(tp+fp):.4f} - Wie oft ist Betrugs-Alarm korrekt?")
print(f"Recall (Betrug): {tp/(tp+fn):.4f} - Wie viel Betrug wird erkannt?")

# Feature Importance - erstes Level der Erklärbarkeit
print(f"\nSCHRITT 6: FEATURE IMPORTANCE ANALYSE")
print("-" * 40)
feature_importance = rf_model.feature_importances_
feature_importance_dict = dict(zip(feature_names, feature_importance))
sorted_features = sorted(feature_importance_dict.items(), key=lambda x: x[1], reverse=True)

print("TOP 10 WICHTIGSTE FEATURES für Betrugsdetektinn:")
for i, (feature, importance) in enumerate(sorted_features[:10], 1):
    print(f"   {i:2}. {feature}: {importance:.4f}")

# Professionelle Visualisierung
print(f"\nErstelle umfassende Modell-Visualisierung...")

# 1. Plot: Feature Importance
plt.figure(figsize=(10, 8))
top_features = 15
sorted_idx = feature_importance.argsort()[-top_features:]

bars = plt.barh(range(top_features), feature_importance[sorted_idx],
                 color=farben['akzent'], alpha=0.8, edgecolor='black', linewidth=1)

plt.yticks(range(top_features), [feature_names[i] for i in sorted_idx], fontweight='bold', fontsize=9)
plt.xlabel('Feature Importance', fontweight='bold')
plt.title('Top 15 Feature Importance', fontweight='bold', pad=15)
plt.grid(axis='x', alpha=0.3)

for i, bar in enumerate(bars):
    width = bar.get_width()
    plt.text(width + width*0.02, bar.get_y() + bar.get_height()/2,
             f'{feature_importance[sorted_idx[i]]:.3f}',
             ha='left', va='center', fontweight='bold', fontsize=8)

plt.show()


# 2. Plot: Confusion Matrix
plt.figure(figsize=(8, 6))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
            xticklabels=['Legitim', 'Betrug'],
            yticklabels=['Legitim', 'Betrug'],
            cbar_kws={'label': 'Anzahl'})

plt.title('Konfusionsmatrix', fontweight='bold', pad=15)
plt.ylabel('Tatsächliche Klasse', fontweight='bold')
plt.xlabel('Vorhergesagte Klasse', fontweight='bold')
plt.show()


# EU AI Act Compliance Status
print(f"\n" + "="*60)
print("EU AI ACT COMPLIANCE STATUS")
print("="*60)
print("Artikel 13 (Transparenz):")
print("Feature Importance verfügbar")
print("Modell-Performance dokumentiert")
print("Entscheidungsfaktoren identifiziert")
print()
print("Artikel 15 (Genauigkeit):")
print(f"Accuracy: {accuracy:.4f}")
print()
print("Fairness & Bias-Kontrolle:")
print("Class-Weight Balancing implementiert")
print("Unbalance-Problem dokumentiert")
print("Feature-Bias-Check durchgeführt")
print()
print("NÄCHSTE SCHRITTE:")
print("SHAP für globale Erklärbarkeit")
print("LIME für lokale Einzelfall-Erklärungen")
print("Vollständige EU AI Act Dokumentation")

print(f"\nBereit für Explainable AI (XAI)!")
print("="*80)
````

# SHAP: Globale Erklärbarkeit verstehen

## 🎯 **Was ist SHAP?**

**SHAP (SHapley Additive exPlanations)** ist eine Methode für Explainable AI und basiert auf der **Spieltheorie**.

### 🧮 **Die Grundidee: Shapley-Werte**

Stell dir vor, ein **Team von Features** entwickelt gemeinsam eine **Vorhersage**:

#### 👥 **Das Team-Beispiel:**
- **Team-Ziel**: Vorhersage "Ist diese Transaktion Betrug?"
- **Team-Mitglieder**: V1, V2, V3, ... ,V28, Time, Amount (alle Features)
- **Frage**: Wer trägt wie viel zum Endergebnis bei?

**SHAP beantwortet**: *"Wenn wir alle möglichen Team-Kombinationen ausprobieren, wie viel trägt jedes Feature im Durchschnitt bei?"*

---

## ⚖️ **EU AI Act Compliance mit SHAP**

### 📋 **Artikel 13 Anforderungen erfüllen:**

#### ✅ **Transparenz schaffen**
```python
# Beispiel: Warum wurde Transaktion als Betrug klassifiziert?
"Diese Transaktion wurde als Betrug eingestuft aufgrund von:
- V14 (Erhöhte Verdächtigkeit): +0.45 Betrug-Score  
- V4 (Unusual Pattern): +0.23 Betrug-Score
- Amount (Hoher Betrag): +0.12 Betrug-Score
- V10 (Risikomuster): -0.08 Betrug-Score (mindert Verdacht)"
```

---

## 🎯 **SHAP vs. Feature Importance: Der Unterschied**

### 🌳 **Random Forest Feature Importance:**
- ✅ **Schnell**: Kann direkt nach dem Modelltraining ausgegeben werden
- ❌ **Modell-spezifisch**: Kann nicht direkt mit anderen Modellen verglichen werden
- ❌ **Ungenau**: Welche Features sind generell wichtig?

### 🧮 **SHAP-Werte:**
- ✅ **Modellagnostisch**: Kann für beliebige Modelle angewendet werden
- ✅ **Genau**: Exakte Zahlen, nicht nur "wichtig/unwichtig"
- ⚠️ **Rechenaufwand**: Kann bei großen Modellen langsam sein

---

## 🛠️ **SHAP-Analyse für unser Betrugsdetektions-System**

### 📋 **EU AI Act Benefits:**
- **Vollständige Transparenz**: Jede Entscheidung ist nachvollziehbar
- **Praktische Anwendung**: Kundenberatung und Aufsichtsbehörden
- **Bias-Erkennung**: Unerwünschte Muster werden sichtbar

---

Wir werden jetzt **SHAP** in unser Random Forest Modell integrieren und **umfassende Erklärungen** für unser Betrugsdetektions-System erstellen!

````python
import shap
import time
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd

# JavaScript für interaktive Plots aktivieren
shap.initjs()

# Matplotlib-Konfiguration für bessere Plots
plt.rcParams['figure.dpi'] = 100
plt.rcParams['savefig.dpi'] = 100
plt.rcParams['font.size'] = 12

# 🔍 SHAP Analyse - Vollständige Modell-Erklärbarkeit
print("SHAP-ANALYSE: GLOBALE ERKLÄRBARKEIT FÜR EU AI ACT")
print("Interpretierbarkeit für Betrugsdetektinn")
print("="*80)

# SHAP installieren und importieren
try:
    import shap
    print("SHAP bereits installiert")
except ImportError:
    print("📦 Installiere SHAP...")
    import subprocess
    subprocess.run(['pip', 'install', 'shap'], check=True)
    import shap
    print("SHAP erfolgreich installiert")

# SHAP Explainer erstellen
print("\nSCHRITT 1: SHAP EXPLAINER ERSTELLEN")
print("-" * 50)
print("Erstelle TreeExplainer für Random Forest...")
explainer = shap.TreeExplainer(rf_model)
print("TreeExplainer erfolgreich erstellt")

# Stichprobe für Analyse (aus Performance-Gründen)
print("\nSCHRITT 2: DATEN FÜR SHAP-ANALYSE VORBEREITEN")
print("-" * 50)
sample_size = 100
sample_indices = np.random.choice(X_test_scaled.shape[0], sample_size, replace=False)
X_sample = X_test_scaled[sample_indices]
y_sample = y_test.iloc[sample_indices]

# SHAP-Werte berechnen
print("\nSCHRITT 3: SHAP-WERTE BERECHNEN")
print("-" * 50)
print("Berechne SHAP-Werte...")
start_time = time.time()
shap_values = explainer.shap_values(X_sample)
calculation_time = time.time() - start_time
print(f"SHAP-Werte berechnet in {calculation_time:.2f} Sekunden")

# Für Binäre Klassifikation nehmen wir Index 1 (Betrugs-Klasse)
# Korrektur: Sicherstellen, dass shap_values die richtige Form hat
if isinstance(shap_values, list) and len(shap_values) == 2:
    shap_values_fraud = shap_values[1]
    expected_value = explainer.expected_value[1]
# Manchmal gibt TreeExplainer ein (n_samples, n_features, n_classes) Array zurück
elif isinstance(shap_values, np.ndarray) and shap_values.ndim == 3:
    shap_values_fraud = shap_values[:, :, 1]
    expected_value = explainer.expected_value[1]
else:
    # Fallback oder Fehlerbehandlung, falls das Format unerwartet ist
    shap_values_fraud = shap_values
    expected_value = explainer.expected_value
    print("Unerwartetes Format der SHAP-Werte. Die Analyse könnte fehlschlagen.")

# Erstelle DataFrame für SHAP-Plots
X_sample_df = pd.DataFrame(X_sample, columns=feature_names)

# Globale Feature-Wichtigkeit mit SHAP
print("\nSCHRITT 4: GLOBALE FEATURE-WICHTIGKEIT")
print("-" * 50)
# Mean absolute SHAP values für globale Wichtigkeit
mean_shap_values = np.abs(shap_values_fraud).mean(axis=0)
feature_importance_shap = pd.DataFrame({
    'feature': feature_names,
    'importance': mean_shap_values
}).sort_values('importance', ascending=False)

print("TOP 10 FEATURES nach SHAP-Analyse:")
print(feature_importance_shap.head(10))


# Umfassende SHAP-Visualisierung
print("\nSCHRITT 5: SHAP-VISUALISIERUNG")
print("-" * 50)

# 1. SHAP Summary Plot (Beeswarm)
print("Erstelle SHAP Summary Plot (Beeswarm)...")
plt.title('SHAP Summary Plot (Beeswarm)', fontsize=16, fontweight='bold', pad=20)
shap.summary_plot(shap_values_fraud, X_sample_df, max_display=15, show=True)
print("SHAP Summary Plot erfolgreich erstellt")


# EU AI Act Compliance-Bericht für SHAP
print("\n" + "="*70)
print("📋 EU AI ACT COMPLIANCE: SHAP-ANALYSE VOLLSTÄNDIG")
print("="*70)
print("Artikel 13 (Transparenz und Information) erfüllt:")
print("Exakte Shapley-Werte berechnet")
print("Globale Feature-Wichtigkeit dokumentiert")
print("Standardisierte Visualisierungen für klare Verständlichkeit")
print()
print("Artikel 15 (Genauigkeit und Robustheit) erfüllt:")
print("Wissenschaftlich fundierte Erklärbarkeit")
print("Axiomatisch bewiesene Fairness-Eigenschaften")
print()
print("STATUS: SHAP für wissenschaftliche EU AI Act Compliance!")
print("Nächster Schritt: LIME für benutzerfreundliche Erklärungen!")
print("="*80)
````

#### 🔍 **Die Grundidee: Lokale Approximation**

  - ✅ **Ungewöhnliche Uhrzeit** (+0.2 Verdacht)  
  - ✅ **Standort abweichend** (+0.15 Verdacht)
  - ❌ **Aber: Bekannter Händler** (-0.1 Verdacht)

**LIME macht komplexe Modelle verständlich für Menschen ohne ML-Hintergrund!**

## 🛠️ **LIME-Funktionsweise: Schritt für Schritt**

### 🎯 **Der LIME-Prozess:**

1. **🎯 Ziel-Instanz auswählen**: Eine spezielle Transaktion
2. **🔄 Samples generieren**: Ähnliche Transaktionen erstellen
3. **📊 Modell befragen**: Vorhersagen für alle Samples
4. **📈 Lineares Modell trainieren**: Einfache Approximation
5. **🔍 Erklärung extrahieren**: Top-Features und ihre Gewichte

````python
import lime
import lime.lime_tabular
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd

# 🔍 LIME Analyse - Lokale Erklärbarkeit für Endnutzer
print("LIME-ANALYSE: LOKALE ERKLÄRBARKEIT FÜR ENDNUTZER")
print("Kundenfreundliche Interpretierbarkeit für EU AI Act Compliance")
print("="*80)

# LIME installieren und importieren
try:
    import lime
    import lime.lime_tabular
    print("LIME bereits installiert")
except ImportError:
    print("Installiere LIME...")
    import subprocess
    subprocess.run(['pip', 'install', 'lime'], check=True)
    import lime
    import lime.lime_tabular
    print("LIME erfolgreich installiert")

# LIME Explainer erstellen
print("\nSCHRITT 1: LIME EXPLAINER KONFIGURIEREN")
print("-" * 50)
print("Erstelle LimeTabularExplainer...")
print("Typ: Tabular Data (Kreditkarten-Transaktionen)")
print("Modus: Classification (Betrug vs. Normal)")
print("Sampling: Gaussian (kontinuierliche Features)")

explainer_lime = lime.lime_tabular.LimeTabularExplainer(
    X_train_scaled,
    feature_names=feature_names,
    class_names=['Legitim', 'Betrug'],
    mode='classification',
    discretize_continuous=False,
    sample_around_instance=True,
    random_state=42
)

print("LimeTabularExplainer erfolgreich erstellt")

# Interessante Fälle für Analyse auswählen
print("\nSCHRITT 2: INTERESSANTE FÄLLE AUSWÄHLEN")
print("-" * 50)

# Betrugs-Wahrscheinlichkeiten für Testdaten
betrug_proba = rf_model.predict_proba(X_test_scaled)[:, 1]

# Verschiedene Kategorien definieren
interessante_faelle = {
    'hohe_betrug_wahrscheinlichkeit': np.where((betrug_proba > 0.8) & (y_test == 1))[0],
    'grenzfall_betrug': np.where((betrug_proba > 0.3) & (betrug_proba < 0.7))[0],
    'falsch_positiv': np.where((betrug_proba > 0.5) & (y_test == 0))[0],
    'sichere_normal': np.where((betrug_proba < 0.2) & (y_test == 0))[0]
}

print("Interessante Fälle identifiziert:")
for kategorie, indices in interessante_faelle.items():
    print(f"   📊 {kategorie}: {len(indices)} Fälle")

# LIME-Erklärungen für verschiedene Fälle generieren
print(f"\nSCHRITT 3: LIME-ERKLÄRUNGEN GENERIEREN")
print("-" * 50)

# Fälle für detaillierte Analyse auswählen
faelle = {
    'ECHTE BETRUGS-TRANSAKTION': interessante_faelle['hohe_betrug_wahrscheinlichkeit'][0] if len(interessante_faelle['hohe_betrug_wahrscheinlichkeit']) > 0 else interessante_faelle['grenzfall_betrug'][0],
    'GRENZFALL (unsicher)': interessante_faelle['grenzfall_betrug'][0] if len(interessante_faelle['grenzfall_betrug']) > 0 else interessante_faelle['sichere_normal'][0],
    'FALSCH-POSITIV': interessante_faelle['falsch_positiv'][0] if len(interessante_faelle['falsch_positiv']) > 0 else interessante_faelle['grenzfall_betrug'][1],
    'SICHERE NORMALE TRANSAKTION': interessante_faelle['sichere_normal'][0]
}

# Speichere LIME Daten für Visualisierung
lime_data = []

for fall_name, idx in faelle.items():
    print(f"\nFall {len(lime_data)+1}: {fall_name}")
    print("-" * 40)
    
    # LIME Erklärung generieren
    print("⏳ Generiere LIME-Erklärung...")
    explanation = explainer_lime.explain_instance(
        X_test_scaled[idx], 
        rf_model.predict_proba, 
        num_features=8,
        top_labels=2
    )
    
    # Vorhersage und tatsächliche Klasse
    vorhersage = rf_model.predict([X_test_scaled[idx]])[0]
    wahrscheinlichkeiten = rf_model.predict_proba([X_test_scaled[idx]])[0]
    tatsaechlich = y_test.iloc[idx]
    
    # Ergebnisse anzeigen
    vorhersage_emoji = "BETRUG" if vorhersage == 1 else "LEGITIM"
    tatsaechlich_emoji = "BETRUG" if tatsaechlich == 1 else "LEGITIM"
    korrekt = "KORREKT" if vorhersage == tatsaechlich else "FALSCH"
    
    print(f"VORHERSAGE: {vorhersage_emoji}")
    print(f"Betrugs-Wahrscheinlichkeit: {wahrscheinlichkeiten[1]*100:.1f}%")
    print(f"Legitim-Wahrscheinlichkeit: {wahrscheinlichkeiten[0]*100:.1f}%")
    print(f"Tatsächlich: {tatsaechlich_emoji}")
    print(f"{korrekt}")
    
    # LIME Features extrahieren
    lime_features = explanation.as_list(label=1)
    
    print(f"\nLIME-ERKLÄRUNG - Warum {vorhersage_emoji}?")
    print("Feature-Beiträge (höhere Werte = mehr Betrugs-Verdacht):")
    
    for feature, weight in lime_features:
        emoji = "🔴" if weight > 0 else "🟢"
        einfluss = "erhöht Betrugs-Verdacht" if weight > 0 else "verringert Betrugs-Verdacht"
        print(f"   {emoji} {feature}: {weight:+.3f} ({einfluss})")
    
    # Für Visualisierung speichern
    lime_data.append({
        'name': fall_name,
        'explanation': explanation,
        'idx': idx,
        'prediction': vorhersage,
        'actual': tatsaechlich,
        'prob_fraud': wahrscheinlichkeiten[1],
        'lime_features': lime_features
    })

# Professionelle LIME-Visualisierung erstellen
print(f"\nSCHRITT 4: LIME-VISUALISIERUNG")
print("-" * 50)
print("Erstelle benutzerfreundliche LIME-Visualisierungen...")

# Verbesserte Subplot-Konfiguration mit mehr Platz
fig, axes = plt.subplots(2, 2, figsize=(22, 16))
fig.suptitle('LIME-Erklärungen: Lokale Interpretierbarkeit (EU AI Act)', 
             fontsize=20, fontweight='bold', y=0.96)

# Farben für bessere Unterscheidung
colors = {
    'positive': '#e74c3c',  # Rot für Betrugs-Verdacht
    'negative': '#27ae60',  # Grün für Anti-Betrugs-Evidenz
    'neutral': '#95a5a6'    # Grau für neutrale Features
}

for i, data in enumerate(lime_data):
    if i >= 4:  # Nur 4 Plots
        break
        
    row, col = i // 2, i % 2
    ax = axes[row, col]
    
    # Feature-Namen und Gewichte extrahieren
    features = []
    weights = []
    
    for feature, weight in data['lime_features']:
        # Feature-Namen kürzen für bessere Lesbarkeit
        if len(feature) > 15:
            short_feature = feature[:12] + "..."
        else:
            short_feature = feature
        features.append(short_feature)
        weights.append(weight)
    
    # Farben basierend auf Gewichten
    bar_colors = [colors['positive'] if w > 0 else colors['negative'] for w in weights]
    
    # Horizontaler Balkenplot mit verbesserter Formatierung
    bars = ax.barh(range(len(features)), weights, color=bar_colors, alpha=0.8, edgecolor='black', linewidth=1)
    
    # Achsen-Konfiguration
    ax.set_yticks(range(len(features)))
    ax.set_yticklabels(features, fontsize=9, fontweight='bold')
    ax.set_xlabel('LIME Weight (Betrugs-Verdacht)', fontweight='bold', fontsize=11)
    
    # Titel mit Falltyp und Emoji
    fall_typ = data['name'].split()[-1] if data['name'] else "Unbekannt"
    prob_text = f"{data['prob_fraud']*100:.1f}%"
    status = "✓" if data['prediction'] == data['actual'] else "✗"
    
    ax.set_title(f"{fall_typ}\n{prob_text} Betrugs-Wahrscheinlichkeit ({status})", 
                fontweight='bold', fontsize=12, pad=15)
    
    # Gewichtswerte auf Balken anzeigen - verbesserte Positionierung
    for bar, weight in zip(bars, weights):
        width = bar.get_width()
        x_pos = width + (max(weights) - min(weights)) * 0.01 if width >= 0 else width - (max(weights) - min(weights)) * 0.01
        ax.text(x_pos, bar.get_y() + bar.get_height()/2, 
               f'{weight:.3f}', 
               ha='left' if width >= 0 else 'right', 
               va='center', fontweight='bold', fontsize=9)
    
    # Grid für bessere Lesbarkeit
    ax.grid(axis='x', alpha=0.4, linestyle='--')
    ax.axvline(x=0, color='black', linewidth=1, alpha=0.8)
    
    # Y-Achse umkehren für bessere Lesbarkeit (wichtigste Features oben)
    ax.invert_yaxis()

# Verbesserte Subplot-Abstände
plt.subplots_adjust(left=0.08, bottom=0.08, right=0.95, top=0.90, wspace=0.3, hspace=0.4)
plt.show()

# EU AI Act Compliance-Bericht für LIME
print(f"\n" + "="*70)
print("EU AI ACT COMPLIANCE: LIME-ANALYSE VOLLSTÄNDIG")
print("="*70)
print("Artikel 13 (Nutzertransparenz) erfüllt:")
print("Benutzerfreundliche Einzelfall-Erklärungen")
print("Verständliche Sprache ohne Fachbegriffe")
print("Quantifizierte Feature-Beiträge")
print("Lokale Entscheidungslogik erklärt")
print()
print("Artikel 50 (Informationspflichten) erfüllt:")
print("KI-Kennzeichnung: Nutzer wissen, es ist KI")
print("Transparente Erklärung: Warum diese Entscheidung?")
print("Nachvollziehbar: Basis für Kundengespräche")
print()
print("Praktische Anwendung:")
print("UI-Integration: Schnelle, intuitive Erklärungen")
print("Kundenservice: Begründung für Sperren/Freigaben")
print("Aufsichtsbehörden: Einzelfall-Nachweise")
print("Entwicklung: Modell-Debugging und -Verbesserung")
print()
print("STATUS: LIME für endnutzer-freundliche EU AI Act Compliance!")
print()
print("Bereit für finale Compliance-Bewertung!")
print("="*80)
````
